```python
# 1. Import all required Python libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 2. Load dataset (download Iris.csv from Kaggle and place in same folder as notebook)
df = pd.read_csv("Iris.csv")

# Display first 5 rows
df.head()
```

```python
# 3. Description of dataset
print("Dataset Description:")
print("""
- SepalLengthCm: Sepal length in centimeters
- SepalWidthCm: Sepal width in centimeters
- PetalLengthCm: Petal length in centimeters
- PetalWidthCm: Petal width in centimeters
- Species: Species of Iris (Setosa, Versicolor, Virginica)
""")

print("Source: https://www.kaggle.com/datasets/uciml/iris")
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Summary statistics
print("\nStatistical summary:")
print(df.describe())
```

```python
# Check shape (rows, columns)
print("\nDataset dimensions:", df.shape)
```

```python
# Check data types and nulls
print("\nData info:")
df.info()
```

```python
# 5. Data Formatting

# Drop the 'Id' column as it is not useful
df = df.drop(columns=['Id'])
```

```python
# Ensure correct data types
df['Species'] = df['Species'].astype('category')
```

```python
# Manual min-max normalization of numeric columns
numeric_cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
for col in numeric_cols:
    df[col] = (df[col] - df[col].min()) / (df[col].max() - df[col].min())
```

```python
# Confirm normalization
df.head()
```

```python
# 6. Convert categorical to quantitative

# Convert 'Species' column to numeric using one-hot encoding
df_encoded = pd.get_dummies(df, columns=['Species'])
```

```python
# Show final result
print("\nData after converting categorical to numeric:")
df_encoded.head()
```
