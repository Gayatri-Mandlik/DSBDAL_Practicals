```python
# Step 1: Import necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Step 2: Load the Iris dataset directly using seaborn
iris = sns.load_dataset('iris')  # Built-in dataset, no need to download
```

```python
# Step 3: Show first few rows
print("🔹 First 5 Rows of Dataset:")
print(iris.head())
```

```python

# Step 4: Dataset information
print("\n🔹 Dataset Info:")
print(iris.info())
```

```python
# Step 5: Feature types
print("\n🔹 Feature Types:")
for col in iris.columns:
    dtype = iris[col].dtype
    if dtype == 'object':
        ftype = 'Nominal (Categorical)'
    else:
        ftype = 'Numeric'
    print(f"{col}: {ftype}")
```

```python
# Step 6: Histograms for Numeric Features
numeric_features = iris.select_dtypes(include=['float64'])

print("\n🔹 Plotting Histograms for Numeric Features:")
numeric_features.hist(bins=20, figsize=(12, 8), color='lightblue', edgecolor='black')
plt.suptitle("Histograms for Iris Dataset Features")
plt.tight_layout()
plt.show()
```

```python
# Step 7: Boxplots for Numeric Features
plt.figure(figsize=(12, 8))
for i, col in enumerate(numeric_features.columns, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(y=iris[col], color='orange')
    plt.title(f'Boxplot of {col}')
    plt.grid(True)
plt.tight_layout()
plt.show()
```
