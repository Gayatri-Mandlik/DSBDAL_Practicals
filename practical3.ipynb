```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the Iris dataset
df = sns.load_dataset("Iris")
df.head()
```

```python
# Group by 'species' and describe numeric variables
grouped_stats = df.groupby('species').agg(['mean', 'median', 'min', 'max', 'std'])
grouped_stats
```

```python
# Create a dictionary mapping species to list of sepal lengths
species_sepal_lengths = df.groupby('species')['sepal_length'].apply(list).to_dict()
species_sepal_lengths
```

```python
# Filter dataset by species
setosa = df[df['species'] == 'setosa']
versicolor = df[df['species'] == 'versicolor']
virginica = df[df['species'] == 'virginica']
```

```python
print("Setosa Stats:")
display(setosa.describe(percentiles=[.25, .5, .75]))

print("\nVersicolor Stats:")
display(versicolor.describe(percentiles=[.25, .5, .75]))

print("\nVirginica Stats:")
display(virginica.describe(percentiles=[.25, .5, .75]))
```

```python
# Boxplot of sepal length by species
sns.boxplot(x='species', y='sepal_length', data=df)
plt.title("Sepal Length Distribution by Species")
plt.show()
```

```python
# Boxplot of petal length by species
sns.boxplot(x='species', y='petal_length', data=df)
plt.title("Petal Length Distribution by Species")
plt.show()
```
